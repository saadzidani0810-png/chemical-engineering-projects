# Projet 3 — Coupling Between Seawater Reverse Osmosis and PEM Electrolysis

## Introduction

The objective of this project is to study the coupling between a **seawater reverse osmosis desalination unit** and a **PEM electrolyzer** for large-scale hydrogen production.

The target annual hydrogen production is:

\[
1 \times 10^6 \ \text{tonnes of H}_2/\text{year}
\]

This large hydrogen production requires a continuous supply of high-purity water. Since seawater is available but cannot be used directly in a PEM electrolyzer, a reverse osmosis desalination unit is introduced before the electrolysis system.

The global system is therefore composed of two main parts:

- a **reverse osmosis desalination unit**, used to produce desalinated water;
- a **PEM electrolysis unit**, used to convert water into hydrogen.

The water demand calculated for the project is:

\[
Q = 1020 \ \text{m}^3/\text{h}
\]

The desalination power is calculated using:

\[
P_{desal} = Q \times E_{sp}
\]

where:

- \(Q\) is the water flow rate in \(\text{m}^3/\text{h}\);
- \(E_{sp}\) is the specific energy consumption of desalination in \(\text{kWh/m}^3\).

In this study, two capacity factors are considered:

\[
CF = 80\%
\]

\[
CF = 50\%
\]

The capacity factor represents the effective operating time of the plant during the year. A lower capacity factor means that the system operates for fewer hours, so the installed electrolyzer capacity must increase in order to produce the same annual amount of hydrogen.

Three reverse osmosis recovery ratios are also studied:

\[
RR = 45\%
\]

\[
RR = 50\%
\]

\[
RR = 60\%
\]

The recovery ratio affects the desalination unit. It changes the amount of feed seawater required, the brine concentration, and the specific energy consumption of the RO process.

The purpose of the analysis is to compare:

- the effect of the **capacity factor** on the electrolyzer size;
- the effect of the **current density** on the electrolyzer active area;
- the effect of the **RO recovery ratio** on desalination energy consumption;
- the relative importance of desalination power compared with electrolysis power.

Overall, the study shows that the electrolysis unit dominates the total energy demand, while the desalination unit represents only a very small fraction of the total power consumption.

# Case 1 — CF = 80% and RR = 50%

## Operating Conditions

In this case, the capacity factor is kept at **80%**, while the reverse osmosis recovery ratio is increased to **50%**.

\[
CF = 80\%
\]

\[
RR = 50\%
\]

## Electrolyzer Sizing

The electrolyzer sizing does not change when the recovery ratio changes, because the recovery ratio only affects the desalination unit.

For \(CF = 80\%\), the operating current is:

\[
I = 3.82 \times 10^9 \ \text{A}
\]

At:

\[
j = 0.6 \ \text{A/cm}^2
\]

the active electrolyzer area remains:

\[
A = 6.37 \times 10^5 \ \text{m}^2
\]

The electrolysis power is:

\[
P_{electrolysis} = 8.33 \ \text{GW}
\]

## Desalination Performance

At a recovery ratio of **50%**, the specific energy consumption is:

\[
E_{sp} = 2.35 \ \text{kWh/m}^3
\]

For:

\[
Q = 1020 \ \text{m}^3/\text{h}
\]

the desalination power is:

\[
P_{desal} = 2.40 \ \text{MW}
\]

## Interpretation

This case gives the lowest desalination energy consumption among the studied recovery ratios.

Compared with \(RR = 45\%\), increasing the recovery ratio to \(50\%\) reduces the desalination power from:

\[
2.54 \ \text{MW}
\]

to:

\[
2.40 \ \text{MW}
\]

Therefore, \(RR = 50\%\) is the best recovery ratio from an energy point of view for the desalination unit.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyomo.environ import (ConcreteModel, Var, Reals, Objective, Constraint,
                           Param, TransformationFactory, value, units)
from pyomo.network import Arc

!pip install pint
!pip install idaes-pse --pre
!pip install watertap

from idaes.core import FlowsheetBlock
from idaes.core import UnitModelCostingBlock
from idaes.models.unit_models import Product, Feed
from idaes.core.util.model_statistics import degrees_of_freedom
from idaes.core.solvers import get_solver
from idaes.core.util.scaling import calculate_scaling_factors, set_scaling_factor
from idaes.core.util.scaling import constraint_scaling_transform
from idaes.core.util.initialization import propagate_state
from watertap.unit_models.reverse_osmosis_0D import (ReverseOsmosis0D, ConcentrationPolarizationType,
                                                     MassTransferCoefficient, PressureChangeType)
import watertap.property_models.seawater_prop_pack as Prop_pack
from watertap.unit_models.pressure_changer import Pump, EnergyRecoveryDevice
from watertap.costing import WaterTAPCosting




LCOE_Grid = 0.11
LCOE_RE = 3.49
LCOE_Hybrid = 3.553

CF = 0.80                    # Projet 3 case 1: 80%
Capacity_Factor = 100 * CF
Utilization_Factor = CF
CapacityF_to_90 = 90 - Capacity_Factor

RR = 0.50       # Projet 3 base case: 50% recovery
Qp_avg_m3_h = 1020                # Projet 3 average water demand
Qp_target_m3_h = Qp_avg_m3_h      # use this to match Calcul de Projet 3 table
Qp_target_m3_s = Qp_avg_m3_h / 3600
Qf_m3_s = Qp_target_m3_s / RR
Area = [] 	    # Mmebrane area (m²)
Op_Pressure = []    # RO pump operating pressure (bar)
Permeate = []       # Water production rate (m^3/day)
Energy_Consump = [] # Specific energy consumption (kWh/m^3)
RR_list = []
CF_list = []
FeedFlow_m3_h = []
PermeateFlow_m3_h = []
BrineFlow_m3_h = []
DesalPower_MW = []
SEC_Project = []
SEC_WaterTAP = []

SEC_project_dict = {
    0.45: 2.49,
    0.50: 2.35,
    0.60: 2.39,
}
# Setup the RO simulation model
m = ConcreteModel()
m.fs = FlowsheetBlock(dynamic=False)
m.fs.properties = Prop_pack.SeawaterParameterBlock()
m.fs.costing = WaterTAPCosting()  # costing model

# Create unit models
m.fs.feed = Feed(property_package = m.fs.properties)
m.fs.pump = Pump(property_package = m.fs.properties)
m.fs.RO = ReverseOsmosis0D(property_package = m.fs.properties, has_pressure_change=True,
          pressure_change_type = PressureChangeType.calculated,
          mass_transfer_coefficient = MassTransferCoefficient.calculated,
          concentration_polarization_type = ConcentrationPolarizationType.calculated)
m.fs.ERD = EnergyRecoveryDevice(property_package = m.fs.properties)
m.fs.product = Product(property_package = m.fs.properties)
m.fs.disposal = Product(property_package = m.fs.properties)

# Connect unit models
m.fs.s1 = Arc(source=m.fs.feed.outlet, destination=m.fs.pump.inlet)
m.fs.s2 = Arc(source=m.fs.pump.outlet, destination=m.fs.RO.inlet)
m.fs.s3 = Arc(source=m.fs.RO.permeate, destination=m.fs.product.inlet)
m.fs.s4 = Arc(source=m.fs.RO.retentate, destination=m.fs.ERD.inlet)
m.fs.s5 = Arc(source=m.fs.ERD.outlet, destination=m.fs.disposal.inlet)
TransformationFactory("network.expand_arcs").apply_to(m)

# Unit equipment capital and operating costs
m.fs.pump.work_mechanical[0].setlb(0)
m.fs.pump.costing = UnitModelCostingBlock(flowsheet_costing_block=m.fs.costing)
m.fs.RO.costing = UnitModelCostingBlock(flowsheet_costing_block=m.fs.costing)
m.fs.ERD.costing = UnitModelCostingBlock(flowsheet_costing_block = m.fs.costing,
        costing_method_arguments={"energy_recovery_device_type": "pressure_exchanger"})

# System costing - total investment and operating costs
m.fs.costing.cost_process()
m.fs.costing.add_annual_water_production(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_specific_energy_consumption(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_specific_electrical_carbon_intensity(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_LCOW(m.fs.product.properties[0].flow_vol)
# Scaling
# Set default property values
m.fs.properties.set_default_scaling("flow_mass_phase_comp", 1e-2, index=("Liq", "H2O"))
m.fs.properties.set_default_scaling("flow_mass_phase_comp", 1, index=("Liq", "TDS"))

# Set unit model values
set_scaling_factor(m.fs.pump.control_volume.work, 1e-3)
set_scaling_factor(m.fs.ERD.control_volume.work, 1e-3)
set_scaling_factor(m.fs.RO.area, 1e-5)

# Touch properties used in specifying the model
m.fs.feed.properties[0].flow_vol_phase["Liq"]
m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"]
m.fs.disposal.properties[0].flow_vol_phase["Liq"]
m.fs.disposal.properties[0].mass_frac_phase_comp["Liq", "TDS"]

# Calculate and propagate scaling factors
m.fs.pump.efficiency_pump.fix(0.80)                                    # pump efficiency (-)
calculate_scaling_factors(m)
print("DOF before = ", degrees_of_freedom(m))

# Feed, 4 degrees of freedom
m.fs.feed.properties[0].flow_vol_phase["Liq"].fix(2.9e-2)             # Projet 3 feed flowrate
m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"].fix(0.035)  # TDS mass fraction (-)
m.fs.feed.properties[0].pressure.fix(101325)                           # pressure (Pa)
m.fs.feed.properties[0].temperature.fix(273.15 + 25)                   # temperature (K)

# High pressure pump, 2 degrees of freedom
m.fs.pump.control_volume.properties_out[0].pressure.fix(115e5)          # initial pump outlet pressure: 70 bar

# RO unit, 7 degrees of freedom
m.fs.RO.A_comp.fix(4.2e-12)                                            # membrane water permeability coeff (m/Pa/s)
m.fs.RO.B_comp.fix(3.5e-8)                                             # membrane salt permeability coeff (m/s)
m.fs.RO.recovery_vol_phase[0, "Liq"].fix(RR)                           # volumetric recovery (-)
m.fs.RO.feed_side.velocity[0, 0].fix(0.15)                             # crossflow velocity (m/s) *
m.fs.RO.feed_side.channel_height.fix(1e-3)                             # channel height in membrane stage (m)
m.fs.RO.feed_side.spacer_porosity.fix(0.97)                            # spacer porosity in membrane stage (-)
m.fs.RO.permeate.pressure[0].fix(101325)                               # permeate pressure (Pa)

# Energy recovery device, 2 degrees of freedom
m.fs.ERD.efficiency_pump.fix(0.75)                                     # ERD efficiency (-)
m.fs.ERD.control_volume.properties_out[0].pressure.fix(101325)         # ERD outlet pressure (Pa)

print("DOF after = ", degrees_of_freedom(m))
# initialize unit by unit
solver = get_solver()
solver.solve(m.fs.feed) # solve feed

propagate_state(m.fs.s1) # initialize pump
m.fs.pump.initialize()

propagate_state(m.fs.s2) # initialize RO
m.fs.RO.initialize()

propagate_state(m.fs.s4) # initialize energy recovery device
m.fs.ERD.initialize()

# Propagate to product and disposal
propagate_state(m.fs.s3)
propagate_state(m.fs.s5)

# Unfix decision variables and add bounds
m.fs.costing.electricity_cost.fix(0.01*LCOE_Hybrid)
m.fs.costing.utilization_factor.fix(Utilization_Factor)

if hasattr(m.fs.costing, "electrical_carbon_intensity"):
    m.fs.costing.electrical_carbon_intensity.fix(0.55)

# Removed because not available in standard WaterTAPCosting
# m.fs.costing.Grid_Ensured_Fraction.fix(0.01*CapacityF_to_90)
# m.fs.costing.carbon_tax.fix(0.07)
# operating pressure + membrane area + crossflow velocity (i.e. width) - specified water recovery
# pump pressure
m.fs.pump.control_volume.properties_out[0].pressure.unfix()
m.fs.pump.control_volume.properties_out[0].pressure.setlb(40e5)
m.fs.pump.control_volume.properties_out[0].pressure.setub(100e5)
m.fs.pump.deltaP.setlb(0)

# RO crossflow velocity
m.fs.RO.feed_side.velocity[0, 0].unfix()
m.fs.RO.feed_side.velocity.setlb(0.01)
m.fs.RO.feed_side.velocity.setub(1)
m.fs.RO.area.setlb(1)
m.fs.RO.area.setub(5000000)

m.fs.feed.properties[0].flow_vol_phase["Liq"].fix(Qf_m3_s) # Projet 3 feed flowrate (m^3/s)
print("DOF before final solve = ", degrees_of_freedom(m))
results = solver.solve(m)
Vol_Recovery_Rate = 100*value(m.fs.RO.recovery_vol_phase[0, "Liq"]) #%
Mass_Recovery_Rate = 100*value(m.fs.RO.recovery_mass_phase_comp[0, "Liq", "H2O"]) #%
SEC = value(m.fs.costing.specific_energy_consumption)     #kWh/m^3

Feed_mass_flow = sum(m.fs.feed.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS']) # kg/s
Product_mass_flow = sum(m.fs.product.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS'])   # kg/s
Disposal_mass_flow = sum(m.fs.disposal.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS']) # kg/s
Feed_vol_flow = value(units.convert(m.fs.feed.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr))  # m^3/h
Feed_TDS =  value(m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"]) * 1e6 #ppm

Water_prod_flow = value(units.convert(m.fs.product.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr))
Product_TDS = value(m.fs.product.properties[0].mass_frac_phase_comp["Liq", "TDS"]) * 1e6 #ppm

Brine_vol_flow = value(units.convert(m.fs.disposal.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr)) # m^3/h
Brine_TDS = value(m.fs.disposal.properties[0].mass_frac_phase_comp["Liq", "TDS"])* 1e6  #ppm

Operating_pressure = value(units.convert(m.fs.pump.control_volume.properties_out[0].pressure, to_units=units.bar)) # bar
Membrane_area = value(m.fs.RO.area) # m²
Inlet_crossflow_velocity_SW = value(units.convert(m.fs.RO.feed_side.velocity[0, 0], to_units=units.cm / units.s))  # cm/s
Outlet_crossflow_velocity_SW = value(units.convert(m.fs.RO.feed_side.velocity[0, 1], to_units=units.cm / units.s)) # cm/s

Pump_Power = value(units.convert(m.fs.pump.work_mechanical[0], to_units=units.kW))
Power_Recovered = -1 * m.fs.ERD.work_mechanical[0].value / 1e3
ERD_Recovered_Power_SW = -value(units.convert(m.fs.ERD.work_mechanical[0], to_units=units.kW)) # kW
Energy_Flow = Pump_Power - Power_Recovered # kW
Eta_ERD = m.fs.ERD.efficiency_pump[0].value
Water_flux = value(units.convert(m.fs.RO.flux_mass_phase_comp_avg[0,"Liq","H2O"]/(1000*units.kg/units.m**3), to_units=units.mm/units.hr))
Pressure_Drop = -value(units.convert(m.fs.RO.deltaP[0],to_units=units.bar))

Max_interfacial_salinity = value(m.fs.RO.feed_side.properties_interface[0, 1].mass_frac_phase_comp["Liq", "TDS"])*1e6
Feed_density = m.fs.RO.feed_side.properties_out[0].dens_mass_phase['Liq'].value
Feed_viscosity = m.fs.RO.feed_side.properties_out[0].visc_d_phase['Liq'].value
Feed_mass_conc_out = m.fs.RO.feed_side.properties_out[0.0].conc_mass_phase_comp['Liq','TDS'].value

Retentate_mass_flow = sum(m.fs.RO.retentate.flow_mass_phase_comp[0, "Liq", j].value for j in ["H2O", "TDS"])
Retentate_TDS_mass_flow = m.fs.RO.retentate.flow_mass_phase_comp[0, "Liq", "TDS"].value
Retentate_Temperature = m.fs.RO.retentate.temperature[0].value
Retentate_Pressure = m.fs.RO.retentate.pressure[0].value

Brine_mass_flow = sum(m.fs.disposal.inlet.flow_mass_phase_comp[0, "Liq", j].value for j in ["H2O", "TDS"])
Brine_TDS_mass_flow = m.fs.disposal.inlet.flow_mass_phase_comp[0, "Liq", "TDS"].value
Brine_Temperature = m.fs.disposal.inlet.temperature[0].value
Brine_Pressure = m.fs.disposal.inlet.pressure[0].value

SEC = value(m.fs.costing.specific_energy_consumption)     #kWh/m^3
"Electricity requirement of RO unit during 1 year"
Base_period = value(units.convert(m.fs.costing.base_period, to_units=units.hr)) # 1 year = 24*365 hour
# m.fs.costing.aggregate_flow_electricity.fix(To_RO_System)
RO_Electricity_Req = value(m.fs.costing.aggregate_flow_electricity)*Utilization_Factor*Base_period

Permeate_year = Water_prod_flow*value(m.fs.costing.utilization_factor)*Base_period
SEC_project = SEC_project_dict[RR]
P_desal_project_MW = Qp_target_m3_h * SEC_project / 1000
P_desal_WaterTAP_MW = Water_prod_flow * SEC / 1000
Area.append(round(Membrane_area, 2))
Op_Pressure.append(round(Operating_pressure, 2))
Permeate.append(round(Permeate_year, 1))
Energy_Consump.append(round(SEC, 2))
RR_list.append(RR)
CF_list.append(CF)
FeedFlow_m3_h.append(round(Feed_vol_flow, 2))
PermeateFlow_m3_h.append(round(Water_prod_flow, 2))
BrineFlow_m3_h.append(round(Brine_vol_flow, 2))
SEC_Project.append(round(SEC_project, 2))
SEC_WaterTAP.append(round(SEC, 2))
DesalPower_MW.append(round(P_desal_project_MW, 2))

print("________________Projet 3 RO metrics____________________")
print("CF: %.2f" % CF)
print("RR: %.2f" % RR)
print("Feed flowrate: %.2f m3/h" % Feed_vol_flow)
print("Permeate flowrate: %.2f m3/h" % Water_prod_flow)
print("Target permeate flowrate: %.2f m3/h" % Qp_target_m3_h)
print("Brine flowrate: %.2f m3/h" % Brine_vol_flow)
print("Vol Recovery Ratio: %.2f %%" % Vol_Recovery_Rate)
print("Mass Recovery Ratio: %.2f %%" % Mass_Recovery_Rate)
print("Operating pressure: %.2f bar" % Operating_pressure)
print("Membrane area: %.2f m2" % Membrane_area)
print("WaterTAP SEC: %.2f kWh/m3" % SEC)
print("Projet 3 SEC: %.2f kWh/m3" % SEC_project)
print("Projet 3 desalination power: %.2f MW" % P_desal_project_MW)
print("WaterTAP desalination power: %.2f MW" % P_desal_WaterTAP_MW)
print("Water production per year: %.2f m3/y" % Permeate_year)
print("_______________________________________________________")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyomo.environ import (ConcreteModel, Var, Reals, Objective, Constraint,
                           Param, TransformationFactory, value, units)
from pyomo.network import Arc

!pip install pint
!pip install idaes-pse --pre
!pip install watertap

from idaes.core import FlowsheetBlock
from idaes.core import UnitModelCostingBlock
from idaes.models.unit_models import Product, Feed
from idaes.core.util.model_statistics import degrees_of_freedom
from idaes.core.solvers import get_solver
from idaes.core.util.scaling import calculate_scaling_factors, set_scaling_factor
from idaes.core.util.scaling import constraint_scaling_transform
from idaes.core.util.initialization import propagate_state
from watertap.unit_models.reverse_osmosis_0D import (ReverseOsmosis0D, ConcentrationPolarizationType,
                                                     MassTransferCoefficient, PressureChangeType)
import watertap.property_models.seawater_prop_pack as Prop_pack
from watertap.unit_models.pressure_changer import Pump, EnergyRecoveryDevice
from watertap.costing import WaterTAPCosting




LCOE_Grid = 0.11
LCOE_RE = 3.49
LCOE_Hybrid = 3.553

CF = 0.80                    # Projet 3 case 1: 80%
Capacity_Factor = 100 * CF
Utilization_Factor = CF
CapacityF_to_90 = 90 - Capacity_Factor

RR = 0.45      # Projet 3 base case: 50% recovery
Qp_avg_m3_h = 1020                # Projet 3 average water demand
Qp_target_m3_h = Qp_avg_m3_h      # use this to match Calcul de Projet 3 table
Qp_target_m3_s = Qp_avg_m3_h / 3600
Qf_m3_s = Qp_target_m3_s / RR
Area = [] 	    # Mmebrane area (m²)
Op_Pressure = []    # RO pump operating pressure (bar)
Permeate = []       # Water production rate (m^3/day)
Energy_Consump = [] # Specific energy consumption (kWh/m^3)
RR_list = []
CF_list = []
FeedFlow_m3_h = []
PermeateFlow_m3_h = []
BrineFlow_m3_h = []
DesalPower_MW = []
SEC_Project = []
SEC_WaterTAP = []

SEC_project_dict = {
    0.45: 2.49,
    0.50: 2.35,
    0.60: 2.39,
}
# Setup the RO simulation model
m = ConcreteModel()
m.fs = FlowsheetBlock(dynamic=False)
m.fs.properties = Prop_pack.SeawaterParameterBlock()
m.fs.costing = WaterTAPCosting()  # costing model

# Create unit models
m.fs.feed = Feed(property_package = m.fs.properties)
m.fs.pump = Pump(property_package = m.fs.properties)
m.fs.RO = ReverseOsmosis0D(property_package = m.fs.properties, has_pressure_change=True,
          pressure_change_type = PressureChangeType.calculated,
          mass_transfer_coefficient = MassTransferCoefficient.calculated,
          concentration_polarization_type = ConcentrationPolarizationType.calculated)
m.fs.ERD = EnergyRecoveryDevice(property_package = m.fs.properties)
m.fs.product = Product(property_package = m.fs.properties)
m.fs.disposal = Product(property_package = m.fs.properties)

# Connect unit models
m.fs.s1 = Arc(source=m.fs.feed.outlet, destination=m.fs.pump.inlet)
m.fs.s2 = Arc(source=m.fs.pump.outlet, destination=m.fs.RO.inlet)
m.fs.s3 = Arc(source=m.fs.RO.permeate, destination=m.fs.product.inlet)
m.fs.s4 = Arc(source=m.fs.RO.retentate, destination=m.fs.ERD.inlet)
m.fs.s5 = Arc(source=m.fs.ERD.outlet, destination=m.fs.disposal.inlet)
TransformationFactory("network.expand_arcs").apply_to(m)

# Unit equipment capital and operating costs
m.fs.pump.work_mechanical[0].setlb(0)
m.fs.pump.costing = UnitModelCostingBlock(flowsheet_costing_block=m.fs.costing)
m.fs.RO.costing = UnitModelCostingBlock(flowsheet_costing_block=m.fs.costing)
m.fs.ERD.costing = UnitModelCostingBlock(flowsheet_costing_block = m.fs.costing,
        costing_method_arguments={"energy_recovery_device_type": "pressure_exchanger"})

# System costing - total investment and operating costs
m.fs.costing.cost_process()
m.fs.costing.add_annual_water_production(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_specific_energy_consumption(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_specific_electrical_carbon_intensity(m.fs.product.properties[0].flow_vol)
m.fs.costing.add_LCOW(m.fs.product.properties[0].flow_vol)
# Scaling
# Set default property values
m.fs.properties.set_default_scaling("flow_mass_phase_comp", 1e-2, index=("Liq", "H2O"))
m.fs.properties.set_default_scaling("flow_mass_phase_comp", 1, index=("Liq", "TDS"))

# Set unit model values
set_scaling_factor(m.fs.pump.control_volume.work, 1e-3)
set_scaling_factor(m.fs.ERD.control_volume.work, 1e-3)
set_scaling_factor(m.fs.RO.area, 1e-5)

# Touch properties used in specifying the model
m.fs.feed.properties[0].flow_vol_phase["Liq"]
m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"]
m.fs.disposal.properties[0].flow_vol_phase["Liq"]
m.fs.disposal.properties[0].mass_frac_phase_comp["Liq", "TDS"]

# Calculate and propagate scaling factors
m.fs.pump.efficiency_pump.fix(0.80)                                    # pump efficiency (-)
calculate_scaling_factors(m)
print("DOF before = ", degrees_of_freedom(m))

# Feed, 4 degrees of freedom
m.fs.feed.properties[0].flow_vol_phase["Liq"].fix(2.9e-2)             # Projet 3 feed flowrate
m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"].fix(0.035)  # TDS mass fraction (-)
m.fs.feed.properties[0].pressure.fix(101325)                           # pressure (Pa)
m.fs.feed.properties[0].temperature.fix(273.15 + 25)                   # temperature (K)

# High pressure pump, 2 degrees of freedom
m.fs.pump.control_volume.properties_out[0].pressure.fix(115e5)          # initial pump outlet pressure: 70 bar

# RO unit, 7 degrees of freedom
m.fs.RO.A_comp.fix(4.2e-12)                                            # membrane water permeability coeff (m/Pa/s)
m.fs.RO.B_comp.fix(3.5e-8)                                             # membrane salt permeability coeff (m/s)
m.fs.RO.recovery_vol_phase[0, "Liq"].fix(RR)                           # volumetric recovery (-)
m.fs.RO.feed_side.velocity[0, 0].fix(0.15)                             # crossflow velocity (m/s) *
m.fs.RO.feed_side.channel_height.fix(1e-3)                             # channel height in membrane stage (m)
m.fs.RO.feed_side.spacer_porosity.fix(0.97)                            # spacer porosity in membrane stage (-)
m.fs.RO.permeate.pressure[0].fix(101325)                               # permeate pressure (Pa)

# Energy recovery device, 2 degrees of freedom
m.fs.ERD.efficiency_pump.fix(0.75)                                     # ERD efficiency (-)
m.fs.ERD.control_volume.properties_out[0].pressure.fix(101325)         # ERD outlet pressure (Pa)

print("DOF after = ", degrees_of_freedom(m))
# initialize unit by unit
solver = get_solver()
solver.solve(m.fs.feed) # solve feed

propagate_state(m.fs.s1) # initialize pump
m.fs.pump.initialize()

propagate_state(m.fs.s2) # initialize RO
m.fs.RO.initialize()

propagate_state(m.fs.s4) # initialize energy recovery device
m.fs.ERD.initialize()

# Propagate to product and disposal
propagate_state(m.fs.s3)
propagate_state(m.fs.s5)

# Unfix decision variables and add bounds
m.fs.costing.electricity_cost.fix(0.01*LCOE_Hybrid)
m.fs.costing.utilization_factor.fix(Utilization_Factor)

if hasattr(m.fs.costing, "electrical_carbon_intensity"):
    m.fs.costing.electrical_carbon_intensity.fix(0.55)

# Removed because not available in standard WaterTAPCosting
# m.fs.costing.Grid_Ensured_Fraction.fix(0.01*CapacityF_to_90)
# m.fs.costing.carbon_tax.fix(0.07)
# operating pressure + membrane area + crossflow velocity (i.e. width) - specified water recovery
# pump pressure
m.fs.pump.control_volume.properties_out[0].pressure.unfix()
m.fs.pump.control_volume.properties_out[0].pressure.setlb(40e5)
m.fs.pump.control_volume.properties_out[0].pressure.setub(100e5)
m.fs.pump.deltaP.setlb(0)

# RO crossflow velocity
m.fs.RO.feed_side.velocity[0, 0].unfix()
m.fs.RO.feed_side.velocity.setlb(0.01)
m.fs.RO.feed_side.velocity.setub(1)
m.fs.RO.area.setlb(1)
m.fs.RO.area.setub(5000000)

m.fs.feed.properties[0].flow_vol_phase["Liq"].fix(Qf_m3_s) # Projet 3 feed flowrate (m^3/s)
print("DOF before final solve = ", degrees_of_freedom(m))
results = solver.solve(m)
Vol_Recovery_Rate = 100*value(m.fs.RO.recovery_vol_phase[0, "Liq"]) #%
Mass_Recovery_Rate = 100*value(m.fs.RO.recovery_mass_phase_comp[0, "Liq", "H2O"]) #%
SEC = value(m.fs.costing.specific_energy_consumption)     #kWh/m^3

Feed_mass_flow = sum(m.fs.feed.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS']) # kg/s
Product_mass_flow = sum(m.fs.product.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS'])   # kg/s
Disposal_mass_flow = sum(m.fs.disposal.properties[0].flow_mass_phase_comp["Liq", j].value for j in ['H2O', 'TDS']) # kg/s
Feed_vol_flow = value(units.convert(m.fs.feed.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr))  # m^3/h
Feed_TDS =  value(m.fs.feed.properties[0].mass_frac_phase_comp["Liq", "TDS"]) * 1e6 #ppm

Water_prod_flow = value(units.convert(m.fs.product.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr))
Product_TDS = value(m.fs.product.properties[0].mass_frac_phase_comp["Liq", "TDS"]) * 1e6 #ppm

Brine_vol_flow = value(units.convert(m.fs.disposal.properties[0].flow_vol_phase["Liq"], to_units=units.m**3/units.hr)) # m^3/h
Brine_TDS = value(m.fs.disposal.properties[0].mass_frac_phase_comp["Liq", "TDS"])* 1e6  #ppm

Operating_pressure = value(units.convert(m.fs.pump.control_volume.properties_out[0].pressure, to_units=units.bar)) # bar
Membrane_area = value(m.fs.RO.area) # m²
Inlet_crossflow_velocity_SW = value(units.convert(m.fs.RO.feed_side.velocity[0, 0], to_units=units.cm / units.s))  # cm/s
Outlet_crossflow_velocity_SW = value(units.convert(m.fs.RO.feed_side.velocity[0, 1], to_units=units.cm / units.s)) # cm/s

Pump_Power = value(units.convert(m.fs.pump.work_mechanical[0], to_units=units.kW))
Power_Recovered = -1 * m.fs.ERD.work_mechanical[0].value / 1e3
ERD_Recovered_Power_SW = -value(units.convert(m.fs.ERD.work_mechanical[0], to_units=units.kW)) # kW
Energy_Flow = Pump_Power - Power_Recovered # kW
Eta_ERD = m.fs.ERD.efficiency_pump[0].value
Water_flux = value(units.convert(m.fs.RO.flux_mass_phase_comp_avg[0,"Liq","H2O"]/(1000*units.kg/units.m**3), to_units=units.mm/units.hr))
Pressure_Drop = -value(units.convert(m.fs.RO.deltaP[0],to_units=units.bar))

Max_interfacial_salinity = value(m.fs.RO.feed_side.properties_interface[0, 1].mass_frac_phase_comp["Liq", "TDS"])*1e6
Feed_density = m.fs.RO.feed_side.properties_out[0].dens_mass_phase['Liq'].value
Feed_viscosity = m.fs.RO.feed_side.properties_out[0].visc_d_phase['Liq'].value
Feed_mass_conc_out = m.fs.RO.feed_side.properties_out[0.0].conc_mass_phase_comp['Liq','TDS'].value

Retentate_mass_flow = sum(m.fs.RO.retentate.flow_mass_phase_comp[0, "Liq", j].value for j in ["H2O", "TDS"])
Retentate_TDS_mass_flow = m.fs.RO.retentate.flow_mass_phase_comp[0, "Liq", "TDS"].value
Retentate_Temperature = m.fs.RO.retentate.temperature[0].value
Retentate_Pressure = m.fs.RO.retentate.pressure[0].value

Brine_mass_flow = sum(m.fs.disposal.inlet.flow_mass_phase_comp[0, "Liq", j].value for j in ["H2O", "TDS"])
Brine_TDS_mass_flow = m.fs.disposal.inlet.flow_mass_phase_comp[0, "Liq", "TDS"].value
Brine_Temperature = m.fs.disposal.inlet.temperature[0].value
Brine_Pressure = m.fs.disposal.inlet.pressure[0].value

SEC = value(m.fs.costing.specific_energy_consumption)     #kWh/m^3
"Electricity requirement of RO unit during 1 year"
Base_period = value(units.convert(m.fs.costing.base_period, to_units=units.hr)) # 1 year = 24*365 hour
# m.fs.costing.aggregate_flow_electricity.fix(To_RO_System)
RO_Electricity_Req = value(m.fs.costing.aggregate_flow_electricity)*Utilization_Factor*Base_period

Permeate_year = Water_prod_flow*value(m.fs.costing.utilization_factor)*Base_period
SEC_project = SEC_project_dict[RR]
P_desal_project_MW = Qp_target_m3_h * SEC_project / 1000
P_desal_WaterTAP_MW = Water_prod_flow * SEC / 1000
Area.append(round(Membrane_area, 2))
Op_Pressure.append(round(Operating_pressure, 2))
Permeate.append(round(Permeate_year, 1))
Energy_Consump.append(round(SEC, 2))
RR_list.append(RR)
CF_list.append(CF)
FeedFlow_m3_h.append(round(Feed_vol_flow, 2))
PermeateFlow_m3_h.append(round(Water_prod_flow, 2))
BrineFlow_m3_h.append(round(Brine_vol_flow, 2))
SEC_Project.append(round(SEC_project, 2))
SEC_WaterTAP.append(round(SEC, 2))
DesalPower_MW.append(round(P_desal_project_MW, 2))

print("________________Projet 3 RO metrics____________________")
print("CF: %.2f" % CF)
print("RR: %.2f" % RR)
print("Feed flowrate: %.2f m3/h" % Feed_vol_flow)
print("Permeate flowrate: %.2f m3/h" % Water_prod_flow)
print("Target permeate flowrate: %.2f m3/h" % Qp_target_m3_h)
print("Brine flowrate: %.2f m3/h" % Brine_vol_flow)
print("Vol Recovery Ratio: %.2f %%" % Vol_Recovery_Rate)
print("Mass Recovery Ratio: %.2f %%" % Mass_Recovery_Rate)
print("Operating pressure: %.2f bar" % Operating_pressure)
print("Membrane area: %.2f m2" % Membrane_area)
print("WaterTAP SEC: %.2f kWh/m3" % SEC)
print("Projet 3 SEC: %.2f kWh/m3" % SEC_project)
print("Projet 3 desalination power: %.2f MW" % P_desal_project_MW)
print("WaterTAP desalination power: %.2f MW" % P_desal_WaterTAP_MW)
print("Water production per year: %.2f m3/y" % Permeate_year)
print("_______________________________________________________")
```

# Case 3 — CF = 80% and RR = 60%

## Operating Conditions

In this case, the capacity factor remains **80%**, but the recovery ratio is increased to **60%**.

\[
CF = 80\%
\]

\[
RR = 60\%
\]

## Electrolyzer Sizing

The electrolyzer sizing remains controlled by the capacity factor and the current density.

For \(CF = 80\%\):

\[
I = 3.82 \times 10^9 \ \text{A}
\]

At:

\[
j = 0.6 \ \text{A/cm}^2
\]

the active electrolyzer area is:

\[
A = 6.37 \times 10^5 \ \text{m}^2
\]

The electrolysis power is:

\[
P_{electrolysis} = 8.33 \ \text{GW}
\]

## Desalination Performance

At a recovery ratio of **60%**, the specific energy consumption is:

\[
E_{sp} = 2.39 \ \text{kWh/m}^3
\]

The desalination power is:

\[
P_{desal} = 2.44 \ \text{MW}
\]

## Interpretation

Increasing the recovery ratio from \(50\%\) to \(60\%\) slightly increases the desalination energy consumption.

This happens because higher recovery produces a more concentrated brine, which increases the osmotic pressure and makes the RO process more energy demanding.

Even with this increase, the desalination power remains negligible compared with the electrolysis power.

# Case 4 — CF = 50% and RR = 45%

## Operating Conditions

In this case, the capacity factor decreases to **50%**, meaning that the electrolyzer operates for fewer hours during the year.

The reverse osmosis recovery ratio is:

\[
RR = 45\%
\]

\[
CF = 50\%
\]

## Electrolyzer Sizing

Because the plant operates for less time, the electrolyzer must be oversized to produce the same annual amount of hydrogen.

For \(CF = 50\%\), the corrected operating current is:

\[
I = 6.12 \times 10^9 \ \text{A}
\]

At:

\[
j = 0.6 \ \text{A/cm}^2
\]

the required active electrolyzer area is:

\[
A = 1.02 \times 10^6 \ \text{m}^2
\]

At:

\[
j = 0.2 \ \text{A/cm}^2
\]

the required active electrolyzer area is:

\[
A = 3.06 \times 10^6 \ \text{m}^2
\]

The electrolysis power is:

\[
P_{electrolysis} = 13.34 \ \text{GW}
\]

for \(j = 0.6 \ \text{A/cm}^2\).

## Desalination Performance

At \(RR = 45\%\), the desalination specific energy consumption is:

\[
E_{sp} = 2.49 \ \text{kWh/m}^3
\]

The desalination power is:

\[
P_{desal} = 2.54 \ \text{MW}
\]

## Interpretation

This case shows the strong effect of the capacity factor.

When the capacity factor decreases from \(80\%\) to \(50\%\), the electrolyzer becomes much larger.  
However, the desalination power remains almost unchanged and still represents a negligible fraction of the total energy demand.

the active electrolyzer area is:

\[
P_{electrolysis} = 13.34 \ \text{GW}

for \(j = 0.6 \ \text{A/cm}^2\).

## Desalination Performance

At \(RR = 50\%\), the specific energy consumption is:

\[
E_{sp} = 2.35 \ \text{kWh/m}^3
\]

The desalination power is:

\[
P_{desal} = 2.40 \ \text{MW}
\]

## Interpretation

This case gives the best desalination performance among the recovery ratios studied.

However, despite the improvement in desalination, the global system remains dominated by the electrolyzer.

The low capacity factor causes a large increase in the required electrolyzer area and installed power.  
Therefore, the main problem is not the RO unit, but the oversizing of the electrolysis unit.

CF = 50\%

the active electrolyzer area is:

\[

At:

\[

the active electrolyzer area is:

The electrolysis power reaches:

\[
P_{electrolysis} = 13.34 \ \text{GW}
\]

for \(j = 0.6 \ \text{A/cm}^2\).

## Desalination Performance

At \(RR = 60\%\), the specific energy consumption is:

\[
E_{sp} = 2.39 \ \text{kWh/m}^3
\]

\[
P_{desal} = 2.44 \ \text{MW}

## Interpretation

Compared with \(RR = 50\%\), the desalination power increases slightly.

This confirms that increasing the recovery ratio beyond \(50\%\) does not necessarily improve energy performance.  
At higher recovery, the brine becomes more concentrated, which increases the osmotic pressure and the pressure required by the RO unit.

The dominant limitation remains the electrolyzer size caused by the low capacity factor.

# Global Comparison and Final Conclusion

## Effect of Capacity Factor

The capacity factor is the most important parameter for the global system sizing.

When the capacity factor decreases from:

\[
CF = 80\%
\]

to:

\[
CF = 50\%
\]

the electrolyzer must be significantly oversized.

At:

\[
j = 0.6 \ \text{A/cm}^2
\]

the active electrolyzer area increases from:

\[
6.37 \times 10^5 \ \text{m}^2
\]

to:

\[
1.02 \times 10^6 \ \text{m}^2
\]

This means that a lower capacity factor increases the installed area, number of stacks, power demand, and capital cost.

## Effect of Recovery Ratio

The recovery ratio mainly affects the desalination unit.

The desalination specific energy consumption is:

\[
RR = 45\% \Rightarrow E_{sp} = 2.49 \ \text{kWh/m}^3
\]

\[
RR = 50\% \Rightarrow E_{sp} = 2.35 \ \text{kWh/m}^3
\]

\[
RR = 60\% \Rightarrow E_{sp} = 2.39 \ \text{kWh/m}^3
\]

The best value is obtained at:

\[
RR = 50\%
\]

because it gives the lowest desalination energy consumption.

## Desalination Compared With Electrolysis

The desalination power is only around:

\[
2.4 - 2.5 \ \text{MW}
\]

while the electrolysis power is in the range of:

\[
7 - 13 \ \text{GW}
\]

Therefore, desalination is negligible compared with electrolysis.

## Final Conclusion

The best compromise is:

\[
CF = 80\%
\]

\[
j = 0.6 \ \text{A/cm}^2
\]

\[
